## Question 2 - Missing values

I will check which variables contain missing values. For the numerical variables with missing info I will replace the missing values with the mean.

```python
mis_datos.isna().sum()
```

```python
columns_with_missing = [
    "EngineSize",
    "Cylinders",
    "Horsepower",
    "MPG_City",
    "MPG_Highway",
    "Weight",
    "Wheelbase",
    "Length"
]

for column in columns_with_missing:
    average = mis_datos[column].mean()
    mis_datos[column] = mis_datos[column].fillna(average)
```

```python
mis_datos.isna().sum()
```

### Interpretation
There were missing values in 8 numerical variables. I replaced the missing values with the mean of each column, and after that there were no missing values.

## Question 3 - Vehicles by manufacturer

I will count the number of vehicles for each manufacturer and identify the manufacturers with the largest and smallest number of vehicles.

```python
make_counts = mis_datos["Make"].value_counts()

make_counts
```

```python
largest_count = make_counts.max()
smallest_count = make_counts.min()

print("Largest number of vehicles:", largest_count)
print(make_counts[make_counts == largest_count])

print("\nSmallest number of vehicles:", smallest_count)
print(make_counts[make_counts == smallest_count])
```

### Interpretation
Toyota has the most cars in the dataset with 28 and Hummer has the least with only 1 vehicle.

## Question 5 - Price and fuel economy summary

I will calculate the highest and lowest suggested retail prices, alsp the average city and highway miles per gallon.

```python
highest_msrp = mis_datos["MSRP"].max()
lowest_msrp = mis_datos["MSRP"].min()
average_city_mpg = mis_datos["MPG_City"].mean()
average_highway_mpg = mis_datos["MPG_Highway"].mean()

print("Highest MSRP:", highest_msrp)
print("Lowest MSRP:", lowest_msrp)
print("Average city MPG:", average_city_mpg)
print("Average highway MPG:", average_highway_mpg)
```

## Question 8 - Price classification

I will create a new variable called opinion based on the suggested retail price of each vehicle.

```python
def classify_price(price):
    if price > 60000:
        return "Expensive"
    elif price < 30000:
        return "Inexpensive"
    else:
        return "Normal"
```

```python
mis_nuevos_datos["opinion"] = mis_nuevos_datos["MSRP"].apply(classify_price)

mis_nuevos_datos.head()
```

### Interpretation
The opinion variable marks the cars based on their MSRP. Cars less than $30,000 are Inexpensive, cars $30,000 to $60,000 are Normal and the cars higher than $60,000 are Expensive.
