```python
import pandas as pd
import statsmodels.api as sm
```

```python
dataset = pd.read_excel("../data/ACT 3-Exercise 1- Estimation of a Linear Regression Model.xlsx", sheet_name="Sheet1")
```

```python
dataset
```

```python
X = dataset[["Market_Return"]]
Y = dataset["Stock_Return"]
```

```python
X = sm.add_constant(X)
```

```python
model = sm.OLS(Y, X).fit()
```

```python
print(model.summary())
```
